```python
print(df["SeniorCitizen"].value_counts()) #seniorcitizen kolonundaki değerlerin sayısını verir. ilk 15 satırda hiç yaşlı müşteri görünmüyo
print(df["TechSupport"].value_counts()) #techsupport kolonundaki değerlerin sayısını verir. techsupport alan müşterilen sayısı az olduğu için
print(df["InternetService"].value_counts()) #internetservice kolonundaki değerlerin sayısını verir. internet servisi almayan müşterilerin sayısı az olduğu için
```

```python
print(df["Churn"].value_counts()) #kaç kişinin ayrıldığına kaç kişinin ayrılmadığına bakıcaz
```

```python
print(df.shape) #kaç satır kaç kolon var onu gösterir
print(df.info()) #dataframe hakkında bilgi verir. kaç satır kaç kolon var, kolonların veri tipleri, boş değer var mı yok mu gibi bilgiler verir 
print(df.isnull().sum()) #null değer var mı diwe kontrol eder
print(df.isnull().all(axis=1).sum()) #boş satır var mı diye kontrol eder
print(df.duplicated().sum())#kendini tekrar eden satırlar var mı, eğer varsa toplam kaç tane bunu öğreniriz
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce") #boş hücre gördüğünde hata vermez o hücreyi Nan yapar 
#bu sütun sayı olması gerekirken boşluk karakterleri nedeniyle str olarak okumuş program
#sayısal analiz yapabilmek için sütunu sayıya çeviricez, 
```

```python
df.info() #totalcharges sütunun tipini değiştirdikten sonra yeniden kontrol ediyoruz
```

```python
"""
total charges sütunu mğşterinin bugüne kadar şirkete ödediği toplam ücreti gösteriyor.
yularıdaki tabloda total charges sütununa baktığımızda anlıyoruz ki 7032 tane non-null 
yani dolu değer var bu da demek oluyor ki 11 tane nan değerimiz var.
bu satırların neden boş olduğunu anlamak için ( mesela bu ay katılan bir müşterinin
totalcharges değeri 0 olabilir veya şirketten ayrılan bir müşterinin totalcharges değeri boş olabilir ) inceliyoruz.

"""
df[df["TotalCharges"].isnull()][["tenure", "TotalCharges", "Churn"]]
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
"""
yukarıdaki grafikte gördük ki nan değerlere sahip olan kişilerin tenure değerleri yani şirkette geçirdikleri
ay değerleri 0 yani hepsi şirkete yeni katılan müşteriler dolayısıyla da bu zamana kadar şirkete ödedikleri bir 
tutar bulunmuyor. bu yüzden de bu nan değerleri 0 ile doldurabiliriz.

"""
df.info() #nan değerleri 0a çevirdikten sonra yeniden kontrol ediyoruzs
```

```python
df["Churn"] = df["Churn"].map({"Yes": 1, "No": 0}) #churn değeri elimizde str olarak duruyor, veriyi analiz ederken bize kolaylık ssağlaması için 1 ( ayrıldı ) ve 0 ( ayrılmadı ) değerine çeviriyoruz.
df.info() #churn değeri sayısala dönmüş mü diye kontrol ediyoruz.
```
